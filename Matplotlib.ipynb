#Data Visualization
Matplotlib



```python
#Data Visualization
# Importing matplotlib


# Introduction to Matplotlib
# Import Library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# First plot
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
# Add Title
plt.plot(x,y)
plt.title('Linear Graph')
plt.show()
```

```python
# Axis Labels
plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

```python
from matplotlib.lines import lineStyles
# Line Style
plt.plot(
    x,
    y,
    linestyle='--',
    marker="o",
    linewidth=2)
plt.show()
```

```python
# Grid
plt.plot(x,y)
plt.grid(True)
plt.show()
```

```python
# Legend
plt.plot(x,
         y,
         label='Sales')
plt.legend()
plt.show()
```

```python
# Scatter Plot

x = np.random.randint(1,100,50)
y = np.random.randint(1,100,50)
plt.scatter(x,y)
plt.show()
```

```python
# Real DataSet
from sklearn.datasets import load_iris
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
    )

plt.scatter(
    df['sepal length (cm)'],
    df['petal length (cm)']
)

plt.xlabel('Sepal Length')

plt.ylabel('Petal Length')

plt.show()
```

```python
# Bar Chart
subjects = ["Python","ML","Dl","AI"]
students = [45,35,30,25]
plt.bar(
    subjects,
    students
)

plt.title("Students per course")
plt.show()
```

```python
# Horizontal Bar Chart
plt.barh(
    subjects,
    students
)

plt.show()
```

```python
# Histogram
# shows data distribution
marks = np.random.randint(
    50,
    100,
    100
    )
plt.hist(marks)
plt.show()
```

```python
# Histogram with Bins
plt.hist(
    marks,
    bins=10
    )
plt.show()
```

```python
# Pie Chart
labels = [
    "Python",
    "ML",
    "DL",
    "AI"
]
sizes = [
    45,
    35,
    30,
    25
]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%'
)
plt.title("Course Distribution")
plt.show()
```

```python
# Box Plot
# Detect Outliers
data = np.random.normal(
    50,10,100
)
plt.boxplot(data)
plt.show()
```

```python
# Plot Customization
# Figure Size
plt.figure(figsize=(12,8))
plt.plot(x,y)
plt.show()
```

```python
# Multiple Lines
x = [1,2,3,4,5]
y1 = [2,4,6,8,10]
y2 = [1,3,5,7,9]

plt.plot(x,y1,label="Sales")
plt.plot(x,y2,label="Profit")
plt.legend()
plt.show()
```
