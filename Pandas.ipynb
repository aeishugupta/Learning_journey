```python
# Module 5
# GroupBy
# Example
data = {
    'Department':['CSE','ECE','CSE','IT','ECE'],
    'Salary':[50000,60000,55000,70000,65000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Average Salary
df.groupby('Department')['Salary'].mean()
```

```python
#Count
df.groupby('Department')['Salary'].count()
```

```python
#Maximum
df.groupby('Department')['Salary'].max()
```

```python
#Multiple Aggregation
df.groupby('Department')['Salary'].agg(['mean','max','min','count'])
```

```python
Student = pd.DataFrame({
    'ID':[1,2,3],
    'Name':['Alice','Bob','Charlie'],
    'Age':[21,20,19]
})
```

```python
Marks = pd.DataFrame({
    'ID':[1,2,3],
    'Subject':['Math','Science','English'],
    'Marks':[90,85,92]
})
```

```python
merged = pd.merge(Student,Marks,on='ID')
print(merged)
```

```python
# Concatenate
pd.concat(
    [Student,Marks]
    ,axis=1)
```

```python
# Pivot table
pd.pivot_table(
    df,
    values='Salary',
    index='Department',
    aggfunc='mean'
)
```

```python
#remove duplicates
df = df.drop_duplicates()
```

```python
#Filter houses with median income > 5
df_filtered = df[df['median_income'] > 5]
print(df_filtered)
```

```python
#Create a new column (Income * 2 )
df_filtered['income_doubled'] = df_filtered['median_income'] * 2
```

```python
#Sort by house value
df_final = df_filtered.sort_values(by='median_house_value', ascending=True)
print(df_final)
```
