#Data Visualization
Matplotlib



```python
#Data Visualization
# Importing matplotlib


# Introduction to Matplotlib
# Import Library
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# First plot
x = [1,2,3,4,5]
y = [2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
# Add Title
plt.plot(x,y)
plt.title('Linear Graph')
plt.show()
```

```python
# Axis Labels
plt.plot(x,y)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.show()
```

```python
from matplotlib.lines import lineStyles
# Line Style
plt.plot(
    x,
    y,
    linestyle='--',
    marker="o",
    linewidth=2)
plt.show()
```

```python
# Grid
plt.plot(x,y)
plt.grid(True)
plt.show()
```

```python
# Legend
plt.plot(x,
         y,
         label='Sales')
plt.legend()
plt.show()
```
